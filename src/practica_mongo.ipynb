```python
import json
from pymongo import MongoClient
```

```python
# Conexion local con Mongo
client=MongoClient("mongodb://localhost:27017/")

# Bases y datos y colecciones
db=client["Recursos_Humanos"]
coleccion=db["Curriculums"]

# Carga del Json
ruta_json="../data/curriculumns.json"

with open (ruta_json, "r", encoding="utf-8") as f:
    datos=json.load(f)

# Insertar datos
resultado=coleccion.insert_many(datos)
print(f"Insertados {len(resultado.inserted_ids)} documentos")
```

```python
# Consultar los documentos
for doc in coleccion.find():
    print(doc)
```

```python
# Bill Gates
bill=coleccion.find_one({"nombre": "Bill Gates"})

if bill:
    print(f"Nombre: {bill.get('nombre')} \n Titular: {bill.get('titular')}")
```

```python
# Linux fundation
query_linux = {"experiencia.empresa": "Linux Foundation"}

for doc in coleccion.find(query_linux):
    print(f"Nombre: {doc.get('nombre')}\nTitular: {doc.get('titular')}\n")
```

Siempre que lo ejecuto aparece repetido porque cuando ejecuto la celda de insert_manu(datos) mongo añade una copia nueva de todos los documentos. Es decir, crea un id nuevo todo el rato y va acumulando los registros.

```python
# Comienzo formacion 1973
query_1973 = {"formacion.inicio": {"$in": [1973, "1973"]}} # para que busque 1973 y "1973", ya que para mongo no es lo mismo, que es sendible con esas cosas. Es solo para un campo

for doc in coleccion.find(query_1973):
    print(f"Nombre: {doc.get('nombre')}")
    for f in doc.get("formacion", []):
        if str(f.get("inicio")) == "1973":
            print(f"  Institución: {f.get('institucion')}")
            print(f"  Inicio: {f.get('inicio')} | Fin: {f.get('fin')}\n")
```

```python
# Formacion entre 1980-1990

# Filtro
query_80_90 = {
    "$or": [ # Lo mismo que el $in de antes, solo que para varias condiciones 
        {"formacion.inicio": {"$gte": 1980, "$lte": 1990}}, # $gte=mayor o igual que 
        {"formacion.inicio": {"$gte": "1980", "$lte": "1990"}} # $lte=menos o igual que
    ]
}

# Extraccion de instituciones
for doc in coleccion.find(query_80_90):
    instituciones = [
        f.get("institucion") for f in doc.get("formacion", [])
        if 1980 <= int(f.get("inicio", 0)) <= 1990
    ]
    print(f"Nombre: {doc.get('nombre')}")
    print(f"Instituciones: {', '.join(filter(None, instituciones))}\n")
```

```python
# Actualizar resumen 
coleccion.update_one(
    {"nombre": "Steve Wozniak"},
    {"$set": {"resumen": "Silicon Valley | Tech Speaker & Legend"}}
)

print("Resumen de Steave Wozniak actualizado con existo")
```

```python
# Eliminar curriculum
coleccion.delete_one({"nombre": "Linus Torvalds"})
print("Curriculum de Linus Torvalds eliminado")
```

```python
# Contabilizar documentos
total=coleccion.count_documents({})
print(f"Total de documentos en la coleccion: {total}")
```

```python
# Crear indice 
coleccion.create_index([("fromacion.fecha_inicio", 1)])

for indice in coleccion.list_indexes():
    print(indice)
```

```python
# Operacion adicional

# En lugar de hace runa busqueda, procesa los documentos por etapas
pipeline=[
    {"$unwind": "$experiencia"}, # Desglosa la lista de experiencia. Divide un documento en dos individuales identicos, pero cada uno con un unico elemneto de experiencia
    {"$group": {"_id": "$experiencia.empresa", "total_puestos": {"$sum": 1}}}, # Agrupa y cuenta. Une todos los documentos con el mismo nombre
    {"$sort": {"total_puestos": -1}} # Ordena los resultados. El -1 es para orden descendente (mayor a menor)
]

# Ejecuta la pipline y lo imprime
for resultado in coleccion.aggregate(pipeline):
    print(resultado)
```
